## 3 Explore daily demand with a list
**30–45 minutes.** A list stores several values in order. Python indexing starts at zero: `values[0]` is the first item. A slice such as `values[:3]` selects the first three items. `sum()`, `len()` and `max()` calculate a total, count and maximum.

````python
example_demand = [10, 15, 20]
print("First day:", example_demand[0])
print("Mean:", sum(example_demand) / len(example_demand))
````

## 4 Apply a reorder rule and debug it
**45–65 minutes.** `if` runs an indented block when a condition is True; `else` handles the alternative. `<=` means less than or equal to, and `==` tests equality. A `for` loop repeats an action for each item. Use a colon and four spaces of indentation.

For this exercise, reorder point = daily demand × lead time + safety stock. Flag a review when stock is **at or below** that point. Safety stock is an assumed buffer, not calculated from a service level here. In operational systems this rule commonly uses inventory position, including orders and backorders; our assumptions make it equal to stock on hand.

````python
for item in [5, 12]:
    if item <= 10:
        print(item, "is at or below 10")
    else:
        print(item, "is above 10")
````

## 6 Build a small inventory table with pandas
**80–100 minutes.** A dictionary stores key–value pairs inside `{}`. Below, each key is a column name and each value is a list. All lists must have the same length. `pd.DataFrame()` turns this dictionary into a labelled table. Selecting `inventory["stock"]` returns one column; arithmetic on columns works row by row.

Data dictionary: `sku` is a product identifier; `stock` and `safety_stock` are units; `unit_cost` is GBP/unit; `daily_demand` is assumed units/day; `lead_time_days` is replenishment time. Inspect the example and then extend it.

````python
inventory = pd.DataFrame({
    "sku": ["A", "B", "C", "D"],
    "stock": [120, 40, 75, 10],
    "unit_cost": [8.50, 15.00, 4.00, 30.00],
    "daily_demand": [22, 12, 10, 5],
    "lead_time_days": [4, 3, 5, 2],
    "safety_stock": [20, 10, 15, 5],
})
print(inventory)
print("Products:", len(inventory))
````

**Your task:** Add `stock_value`, `reorder_point` and `review_order` columns. Use the same formulas as earlier. Print the total stock value and the rows flagged for review. Which items need review, and what additional information would you request before placing orders?

Pattern for a new column: `inventory["new_column"] = ...`

Pattern for filtering: `inventory.loc[inventory["review_order"], ["sku", "stock", "reorder_point"]]`. Here `.loc` selects rows and columns; the Boolean column chooses rows where the value is True.

````python
# Add three calculated columns, then print the total and flagged rows.
inventory = pd.DataFrame({
    "sku": ["A", "B", "C", "D"],
    "stock": [120, 40, 75, 10],
    "unit_cost": [8.50, 15.00, 4.00, 30.00],
    "daily_demand": [22, 12, 10, 5],
    "lead_time_days": [4, 3, 5, 2],
    "safety_stock": [20, 10, 15, 5],
})
print(inventory)
print("Products:", len(inventory))
inventory["stock_value"] = inventory["daily_demand"] * inventory["unit_cost"]
inventory["reorder_point"] = inventory["daily_demand"] * inventory["lead_time_days"] + inventory["safety_stock"]
inventory["review_order"] = inventory["stock"] <= inventory["reorder_point"]
print(inventory)
inventory.loc[inventory["review_order"], ["sku", "stock", "reorder_point"]]
````

## 7 Visualise demand and explain it
**100–115 minutes.** `plt.subplots()` creates a figure (`fig`) and plotting area (`ax`). Labels need meaningful units. Run this worked example before writing your own chart.

````python
fig, ax = plt.subplots(figsize=(6, 3))
ax.bar(["Example A", "Example B"], [10, 15])
ax.set_xlabel("Days")
ax.set_ylabel("Daily Demand (Units)")
ax.set_title("7 Day Demand Trend")
fig.tight_layout()
plt.show()
````
